```python

import random

graph = {
    'A': {'B': 2, 'C': 4},
    'B': {'A': 2, 'C': 1, 'D': 7, 'E': 5},
    'C': {'A': 4, 'B': 1, 'D': 3, 'E': 2},
    'D': {'B': 7, 'C': 3, 'F': 2},
    'E': {'B': 5, 'C': 2, 'F': 3},
    'F': {'D': 2, 'E': 3}
}

alpha = 1
beta = 5
evaporation_rate = 0.1
Q = 100
num_ants = 10
iterations = 50

pheromone = {
    u: {v: 1.0 for v in graph[u]}
    for u in graph
}

def calculate_cost(path):
    cost = 0
    for i in range(len(path) - 1):
        cost += graph[path[i]][path[i + 1]]
    return cost

def select_next_node(current, visited):
    candidates = [
        v for v in graph[current]
        if v not in visited
    ]

    if not candidates:
        return None

    weights = []
    for v in candidates:
        tau = pheromone[current][v] ** alpha
        eta = (1 / graph[current][v]) ** beta
        weights.append(tau * eta)

    return random.choices(
        candidates, weights=weights, k=1
    )[0]

def construct_path(source, destination):
    path = [source]
    current = source

    while current != destination:
        next_node = select_next_node(current, path)

        if next_node is None:
            return None

        path.append(next_node)
        current = next_node

    return path

def update_pheromone(paths):
    for u in pheromone:
        for v in pheromone[u]:
            pheromone[u][v] *= (1 - evaporation_rate)

    for path in paths:
        cost = calculate_cost(path)
        deposit = Q / cost

        for i in range(len(path) - 1):
            u = path[i]
            v = path[i + 1]

            pheromone[u][v] += deposit
            pheromone[v][u] += deposit

def ant_colony_routing(source, destination):
    best_path = None
    best_cost = float('inf')

    for i in range(iterations):
        paths = []

        for ant in range(num_ants):
            path = construct_path(source, destination)

            if path is not None:
                paths.append(path)
                cost = calculate_cost(path)

                if cost < best_cost:
                    best_cost = cost
                    best_path = path.copy()

        if paths:
            update_pheromone(paths)

    return best_path, best_cost

print("ANT COLONY OPTIMIZATION - NETWORK ROUTING")
print("Available nodes:", list(graph.keys()))

source = input("Enter source node: ").upper()
destination = input("Enter destination node: ").upper()

if source not in graph or destination not in graph:
    print("Invalid node!")
elif source == destination:
    print("Source and destination must be different.")
else:
    random.seed(42)

    path, cost = ant_colony_routing(source, destination)

    if path:
        print("\nOptimal route:", " -> ".join(path))
        print("Total routing cost:", cost)
    else:
        print("No route found.")
```
